```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
np.random.seed(42)  # Rastgele sayıların tekrarlanabilir olması için
```

```python
tarihler = pd.date_range(start="2026-02-01", periods=40, freq='B')  # İş günleri için tarih aralığı
fiyatlar = np.linspace(35.0,42.0, 40) + np.random.normal(0, 1.2, 40)  # Lineer artış + rastgele gürültü                        
fiyatlar = np.round(fiyatlar, 2)  # Fiyatları 2 ondalık basamağa yuvarla
```

```python
df_gram_gumus = pd.DataFrame({'Tarih': tarihler, 
                             'Gram_TL': fiyatlar, 
                             "Gunluk_Hacim" : np.random.randint(5000,25000,40)})  # o gün içinde toplam ne kadar gümüşün el değiştirdiğini (alınıp satıldığını) temsil eder
```

```python
df_gram_gumus.loc[10:12, 'Gram_TL'] = np.nan  # Eksik veri ekle
df_gram_gumus.loc[25, 'Gunluk_Hacim'] = np.nan  # Eksik veri ekle
```

```python
print("Ham Gümüş Gram Verisi:")
print(df_gram_gumus.head(15))
```

```python
# (Gram_TL) boşlukları bir önceki günün verisiyle (Forward Fill) dolduruyoruz
df_gram_gumus["Gram_TL"] = df_gram_gumus["Gram_TL"].ffill()  # Gram_TL için ileri doldurma
hacim_medyan = df_gram_gumus["Gunluk_Hacim"].median() # Gunluk_Hacim sütununun medyanını hesapla
df_gram_gumus["Gunluk_Hacim"] = df_gram_gumus["Gunluk_Hacim"].fillna(hacim_medyan)  # Gunluk_Hacim için medyan ile doldurma
print("\nEksik Veriler Doldurulduktan Sonra:")
print(df_gram_gumus.head(15))
```

```python
yuksek_fiyatlar = df_gram_gumus[df_gram_gumus["Gram_TL"] > 40.0] 
print("\n40 TL'nin Üzerindeki Günler:")
print(yuksek_fiyatlar[["Tarih", "Gram_TL"]])

plt.figure(figsize=(12, 6))
sns.lineplot(data=df_gram_gumus, x="Tarih", y="Gram_TL", marker='o')
plt.title("Gümüş Gram Fiyatları (TL) - Şubat 2026")
plt.xlabel("Tarih")
plt.ylabel("Gram Fiyatı (TL)")
plt.xticks(rotation=45) # Tarih etiketlerini döndür
plt.grid()
plt.tight_layout() # Grafik düzenini sıkıştır
plt.show()
```

```python
# Çizim tuvalini (Figure) ve boyutunu ayarlıyoruz (genişlik 10, yükseklik 5)
plt.figure(figsize=(10, 5))

# X eksenine Tarihleri, Y eksenine temizlenmiş Gram_TL fiyatlarını veriyoruz
plt.plot(df_gram_gumus['Tarih'], df_gram_gumus['Gram_TL'], color='gray', linewidth=2, marker='o')

# Grafiği isimlendirip süslüyoruz
plt.title("2026 Gümüş Gram/TL Fiyat Trendi", fontsize=14, fontweight='bold')
plt.xlabel("Tarih")
plt.ylabel("Gram TL Fiyatı")
plt.grid(True, linestyle='--', alpha=0.6) # Arka plana okumayı kolaylaştıran kesik çizgili ızgara ekliyoruz

# Grafiği ekrana bas!
plt.show()
```
